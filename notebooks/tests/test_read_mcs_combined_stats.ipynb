```python
def is_sequential_unique(da):
    # Access the numpy array directly to avoid xarray overhead
    arr = da.values 
    
    # if arr.size < 2:
    #     return True
    
    # np.diff computes arr[i+1] - arr[i] for all i
    # If every difference is exactly 1, then it's sequentially increasing
    # This also guarantees no duplicates (which would have a diff of 0)
    return np.all(np.diff(arr) == 1)
```
